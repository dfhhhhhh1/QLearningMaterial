# Module 5: Quantum Algorithm Exploration

## Description

With circuits in hand, this module surveys the algorithms that motivate the entire field. Shor's algorithm is the reason quantum computing has an RSA problem attached to it; Deutsch-Jozsa is the clean pedagogical case; VQE and QAOA are the near-term variational workhorses for chemistry, materials, and optimization. The emphasis is on what each algorithm buys you and what it costs.

----


## Guiding Questions

Understand the RSA encryption threat from Shor's algorithm, as well as the Deutsch-Jozsa algorithm, VQE for chemistry and material science, and QAOA for optimization.


## Topics Covered

- Deutsch-Jozsa algorithm
- Shor's algorithm and the RSA encryption threat
- Grover's algorithm and amplitude amplification
- VQE for chemistry and materials science
- QAOA for optimization
- Variational algorithms and the hybrid loop
- Observables and expectation values
- Business impacts of quantum algorithms


## Objectives

- Students will explain the threat Shor's algorithm poses to RSA and the timeline caveats around it
- Students will describe the Deutsch-Jozsa speedup and the assumptions it relies on
- Students will describe the structure of a variational algorithm and the role of the classical optimizer
- Students will identify candidate problems for VQE and for QAOA
- Students will compute expectation values of observables from circuit output
