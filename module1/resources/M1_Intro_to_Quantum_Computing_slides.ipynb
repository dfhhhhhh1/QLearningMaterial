## Slide 4: Are quantum computers faster?

![Slide 4](M1_Intro_to_Quantum_Computing_slides_images/Slide4.PNG)

> Short answer: No
> While quantum computers provide exponential speedup:
> They are not “faster”:
> Quantum computers perform operations slower than classical
> ~1 billion floating point additions on a 4GHz CPU assuming 4 clock cycles per addition and no pipelining
> ~3 million single gate operations on a Heron r3 processor (modern QPU)
> Around 333 times slower

## Slide 12: Comparison to a coin

![Slide 12](M1_Intro_to_Quantum_Computing_slides_images/Slide12.PNG)

> Questions for a Coin:
> “Heads or Tails?”
> “Spinning clock-wise or counter-clockwise?”
> “Is the spin pitch-up or pitch down?”
> Note that when you ask one of these questions, the other questions do not make sense.
> The more you know about the “spin” of the coin, the less you know about the “face” the coin is on.

## Slide 13: Exact Coin Representation

![Slide 13](M1_Intro_to_Quantum_Computing_slides_images/Slide13.PNG)

> Superposition
> The state of simultaneous mutually exclusive states:
> “The coin is both heads and tails”
> Frame of reference
> The coin is spinning on the table: neither heads nor tails.
> Phase
> The coin is spinning right or left
> Decoherence
> The coin slows down due to interaction in the environment
