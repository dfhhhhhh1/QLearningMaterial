# Module 1 Practice Quiz — Randomized

This practice quiz draws a **random set of questions** from a larger bank each time you start it. Answer all of them, then click **Submit** to see your **final score** — you do *not* get per-question feedback along the way, so the goal is to actually know the material, not to guess-and-check.

Want another try? Click **New attempt** to re-roll a fresh random set of questions.

The correct answers live only as one-way hashes in `M1_quiz_bank.json`, so they can't be read from the files you cloned.

> **Requires `ipywidgets`** (standard in JupyterHub). The controls appear when the notebook is *run* in a live kernel; a static preview will look empty.


---
### Setup — run this first


```python
import json, os, re, random, hashlib
import ipywidgets as widgets
from IPython.display import display, clear_output

BANK_PATH = "M1_quiz_bank.json"   # same folder as this notebook

with open(BANK_PATH, encoding="utf-8") as _f:
    _BANK = json.load(_f)
SALT = _BANK["salt"]
_QUESTIONS = _BANK["questions"]

def _normalize(s):
    return re.sub(r"[^a-z0-9]", "", str(s).lower())

def _is_correct(q, value):
    got = hashlib.sha256((SALT + _normalize(value)).encode()).hexdigest()
    return got == q["answer_hash"]

def start_quiz(n=9):
    """Draw a fresh random quiz of n questions. Score is shown only after Submit."""
    n = min(n, len(_QUESTIONS))
    shell = widgets.Output()

    def render():
        with shell:
            clear_output(wait=True)
            sample = random.sample(_QUESTIONS, n)
            fields = []          # list of (question, widget)
            widgets_to_show = []
            for i, q in enumerate(sample, 1):
                widgets_to_show.append(widgets.HTML(f"<b>Q{i}.</b> {q['prompt']}"))
                if q["kind"] == "choice":
                    opts = [tuple(o) for o in q["options"]]
                    field = widgets.RadioButtons(options=opts, value=None,
                                                 layout=widgets.Layout(width="auto"))
                else:
                    field = widgets.Text(placeholder="type your answer")
                fields.append((q, field))
                widgets_to_show.append(field)
                widgets_to_show.append(widgets.HTML("<hr style='opacity:0.3'>"))

            submit = widgets.Button(description="Submit & get final score",
                                    button_style="primary",
                                    layout=widgets.Layout(width="220px"))
            reroll = widgets.Button(description="New attempt (re-roll questions)",
                                    layout=widgets.Layout(width="260px"))
            result = widgets.Output()

            def on_submit(_):
                with result:
                    clear_output(wait=True)
                    missing = [i for i, (q, f) in enumerate(fields, 1)
                               if f.value is None or str(f.value).strip() == ""]
                    if missing:
                        print("Answer every question before submitting. Missing: "
                              + ", ".join(f"Q{i}" for i in missing))
                        return
                    score = 0
                    missed = []
                    for i, (q, f) in enumerate(fields, 1):
                        if _is_correct(q, f.value):
                            score += 1
                        else:
                            missed.append(i)
                    # lock the attempt so it can't be turned into trial-and-error
                    for q, f in fields:
                        f.disabled = True
                    submit.disabled = True
                    print(f"Final score: {score} / {len(fields)}")
                    if missed:
                        print("Missed: " + ", ".join(f"Q{i}" for i in missed))
                        print("Answers are not shown. Click 'New attempt' for a fresh set to try again.")
                    else:
                        print("Perfect score! \U0001F389")

            def on_reroll(_):
                render()

            submit.on_click(on_submit)
            reroll.on_click(on_reroll)

            for w in widgets_to_show:
                display(w)
            display(widgets.HBox([submit, reroll]))
            display(result)

    render()
    display(shell)
```

---
## Take the quiz

Run the cell below. Change `n=9` if you want a different number of questions (the bank currently holds more than that).


```python
start_quiz(n=9)
```
