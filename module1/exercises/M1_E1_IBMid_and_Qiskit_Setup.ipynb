
## Step 2

## Step 2

In python, add this line of code. This will connect to the IBM Cloud. Paste your CRN where it says, "YOUR_CRN_HERE". Paste your API token where it says, "YOUR_API_KEY_HERE".

```python
service = QiskitRuntimeService(channel="ibm_cloud", instance="YOUR_CRN_HERE", token="YOUR_API_HERE")
```


## Step 3

Run the following line of code (you may need to add it to a print statement). It returns a list of quantum computers you can access with your API token. If it prints name(s) of quantum computer(s) (example: "ibm_brisbane") then it worked, and you have successfully connected to the API.

```python
service.backends()
```
