# Module 1: Quantum Fundamentals & Environment Setup

## Description

Quantum computing does not replace classical computing; it extends it with a fundamentally different model of information. This module contrasts the classical bit with the qubit, motivates why certain problems are hard for classical machines, and gets your working environment stood up so that every later module runs against real tooling. By the end of the week you will have an IBMid, a working Qiskit install, and a mental model for what a qubit actually is.

----


## Guiding Questions

Translating and contrasting knowledge of classical computing to quantum computing. What is the motivation for quantum, and what actually is a qubit?


## Topics Covered

- Classical computing vs. quantum computing
- Motivation: which problems are hard, and why
- What a qubit is (and what it is not)
- The IBM Quantum Platform and IBM Quantum Lab
- Installing and verifying Qiskit


## Objectives

- Students will contrast the classical bit with the qubit and describe the difference in their own words
- Students will articulate the motivation for quantum computing and name problem classes where it is expected to matter
- Students will create an IBMid and authenticate against the IBM Quantum Platform
- Students will install Qiskit and verify the installation by running a circuit
- Students will become familiar with the Jupyter environment and the submission protocol for this course
