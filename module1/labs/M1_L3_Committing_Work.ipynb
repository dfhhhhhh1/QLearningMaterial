# Submitting Your Work

All work in this course is submitted with **Git**. When you finish an assignment or a quiz, you *commit* your changes (a save point) and *push* them so the instructors and TAs can see your work.

This notebook walks you through the whole process.

## TL/DR

From a **Terminal** in JupyterHub, inside your course folder:

```bash
git add -A
git commit -m "Completed Module 1 assignment"
git push
```

## Step 1: Save your notebooks

Before committing, make sure every notebook you edited is saved: **File ▸ Save Notebook**, or press **Ctrl+S** (**Cmd+S** on Mac). Git only sees changes that are written to disk.

## Step 2: Open a Terminal

In JupyterHub, click **File ▸ New ▸ Terminal** (or the **Terminal** tile on the Launcher). A command-line prompt will open.

## Step 3 Go to your course folder

Change into the course directory (the folder that contains the `module1`, `module2`, … folders):

```bash
cd f26dsaQLearning*
```

Use the actual name of your course folder if it is different.

## Step 4: (Optional, view changes)

```bash
git status
```

This lists the files you have changed since your last commit. Changed files that are not yet staged appear in red.

## Step 5: Stage your changes

Stage everything you have changed:

```bash
git add -A
```

Or stage a single file:

```bash
git add module1/exercises/M1_E1_IBMid_and_Qiskit_Setup.ipynb
```

## Step 6: Commit (make a save point)

```bash
git commit -m "A short message describing what you did"
```

Write a clear message, e.g. `"Finish Module 2 math homework"`. Each commit is a permanent save point you can return to later.

## Step 7: Push (send it in)

```bash
git push
```

This uploads your commits. **Your work is not submitted until you push.**

## Good habits

- **Commit often**: after each notebook or work session, not just at the end.
- **Write meaningful messages**: "Module 3 Qiskit lab" beats "stuff".
- **Push before the deadline**: only pushed work is submitted.

## Getting new modules

Each week you pull the new module with:

```bash
git pull --no-edit upstream master
```

A short merge message is normal. See **[Start_Here](../../Start_Here.ipynb)** for details.

## Troubleshooting

- **"Everything up-to-date" but I made changes**: you probably forgot to `git add` and `git commit` before pushing.
- **Merge conflict**: ask in the Mutual Aid / MS Teams channel; the instructors and TAs can help.
- **Asked for a username / password / token**: follow your course's Git credential instructions, and **never paste a token into a notebook you commit**.
