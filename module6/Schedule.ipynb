# Module 6: Quantum Error Correction and Fault-Tolerant Quantum

## Description

Qubits are fragile, and that fragility is the central engineering problem of the field. This module separates three ideas that are frequently and unhelpfully conflated: error suppression, error mitigation, and error correction. You will work through the repetition code and stabilizers to see how many noisy physical qubits are combined into one reliable logical qubit.

----


## Guiding Questions

What makes qubits so fragile, and what is the difference between error suppression, error mitigation, and error correction? Understand the repetition code, stabilizers, and how physical qubits become logical qubits.


## Topics Covered

- Sources of noise: decoherence, gate error, readout error
- Error suppression vs. error mitigation vs. error correction
- The repetition code
- Stabilizers and syndrome measurement
- Physical qubits vs. logical qubits
- The road to fault tolerance


## Objectives

- Students will describe the primary sources of noise in a quantum device
- Students will distinguish error suppression, error mitigation, and error correction, and say when each applies
- Students will trace how the repetition code detects and corrects a bit-flip error
- Students will explain the relationship between physical and logical qubits
