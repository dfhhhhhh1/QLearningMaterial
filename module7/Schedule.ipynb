# Module 7: Quantum in Practice and Scaling to Utility

## Description

This module is about running real work at real scale. Quantum utility, the point at which a quantum computer does something useful that is hard to verify classically, is a different and nearer claim than quantum advantage, and the distinction matters. You will work through the Qiskit patterns workflow and the Sampler and Estimator primitives, which is how 100+ qubit jobs actually get submitted.

----


## Guiding Questions

What does it take to run a real 100+ qubit job? Understand quantum utility vs. quantum advantage, the Qiskit patterns workflow, and the Sampler and Estimator primitives.


## Topics Covered

- Quantum utility vs. quantum advantage
- The Qiskit patterns workflow: map, optimize, execute, post-process
- The Sampler primitive
- The Estimator primitive
- Mapping problems onto hardware topology
- Running circuits at scale


## Objectives

- Students will distinguish quantum utility from quantum advantage
- Students will apply the four-step Qiskit patterns workflow to a problem of their own
- Students will choose correctly between the Sampler and Estimator primitives for a given task
- Students will map a problem onto a device's qubit topology and explain the cost of that mapping
- Students will implement a full algorithm end to end on hardware
