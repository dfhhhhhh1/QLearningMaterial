Grover's Algorithm


# Objective

The Objective of this assignment is to become familiar with how Grover's algorithm works and learn how to implement the algorithm. 

 

**Grover's algorithm** is a quantum search *algorithm* devised by Lov Grover in 1996. It provides a *quadratic speedup* for searching an unsorted database or, more generally, for solving problems that can be framed as finding one (or more) target items among many possibilities.

The following is a list of commonly claimed uses of Grover's algorithm

- Database Search
- [Traveling Salesman](https://en.wikipedia.org/wiki/Travelling_salesman_problem)
- [3-SAT problem](https://en.wikipedia.org/wiki/Boolean_satisfiability_problem)

Grover's algorithm is also one of the algorithms that (lightly) threatens our cryptographic systems, requiring small updates to the cryptography used in communications, such as requiring larger key sizes.

# Objective

The Objective of this assignment is to become familiar with how QAOA works and learn how to implement the algorithm for solving Max-Cut, while learning some of the potential applications. 

 

**QAOA** (Quantum Approximate Optimization Algorithm) is a variational quantum algorithm that uses a hybrid quantum-classical loop to optimize cost functions and approximate best solutions to complex optimization problems. For this assignment, QAOA will be applied to solving the [Max-Cut problem](https://en.wikipedia.org/wiki/Maximum_cut), which falls under the list of NP-hard problems, which are difficult for classical computers to solve efficiently. Although this specific application is more theoretical, QAOA can be used for an array of combinatorial optimization problems.

 

Here is a link to the [IBM tutorial on QAOA](https://quantum.cloud.ibm.com/docs/en/tutorials/quantum-approximate-optimization-algorithm), which applies it to the Maximum Cut problem. This is also located in the modules. This IBM tutorial provides most of the learning resources that you will need for understanding the basics. There is also this tutorial from IBM: 

[Solving MaxCut with QAOA: Quantum Tutorial for 100+ Qubits](https://www.youtube.com/watch?v=P3s7TIMIvZ0)

 

For people who are interested in reading the original paper introducing the Quantum Approximate Optimization Algorithm by Edward Farhi, Jeffrey Goldstone, and Sam Gutmann, you can [view the PDF on arXiv here](https://arxiv.org/abs/1411.4028).
