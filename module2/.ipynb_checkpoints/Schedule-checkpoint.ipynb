# Module 2: Quantum Mathematics and Notations with Linear Algebra

## Description

Quantum computing is linear algebra with a physical interpretation. This module builds the notation and the mathematical machinery you will use for the rest of the course: Dirac notation for states, unitary matrices for operations, and the Born rule for measurement. Superposition stops being a metaphor here and becomes a vector you can write down.

----


## Guiding Questions

How do I understand Dirac notation of quantum systems, why are unitary matrices important, what is superposition, and what does it mean to measure a quantum state?


## Topics Covered

- Dirac (bra-ket) notation
- State vectors, amplitudes, and normalization
- Unitary matrices and why reversibility matters
- Superposition
- Measurement and the Born rule
- The Bloch sphere
- The IBM Composer


## Objectives

- Students will read and write single-qubit states in Dirac notation
- Students will explain why quantum operations must be unitary
- Students will compute measurement probabilities from a state vector
- Students will locate a single-qubit state on the Bloch sphere
- Students will build and inspect a simple circuit in the IBM Composer
