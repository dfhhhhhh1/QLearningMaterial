# Bonus Material

## Description

Expand on material covered in this course to introduce other quantum machine types (trapped ion, neutral atom, photonic, annealing), future algorithms, and fault-tolerant schemas including qLDPC codes.

----


## Topics Covered

- Trapped-ion quantum computers
- Neutral-atom quantum computers
- Photonic quantum computers
- Quantum annealing
- Fault-tolerant schemas, including qLDPC codes
