# Welcome to Mizzou's QLearning, Quantum Computing with Qiskit
## An 8-Week introduction to bring you from beginner to practitioner, built on IBM Quantum Learning

This course introduces you to quantum computing from the ground up. Over eight weeks, you will
explore the essential mathematical concepts, quantum circuits, and Qiskit programming needed to
work with real quantum hardware, then extend into error correction, utility-scale practice, and
current research.

Two ideas drive how this course is built:<BR>
&nbsp;&nbsp;&nbsp;&nbsp;1) **Quantum computing is learned by running things.** Every module pairs
conceptual material with a notebook you execute yourself. The mathematics is an important early component of this course;
you will use it, in code, against a simulator and against real quantum hardware.<BR>
&nbsp;&nbsp;&nbsp;&nbsp;2) **The hype and the engineering are separable, and you should be able to
separate them.** A great deal is claimed about quantum computing. By the end of this course you
should be able to read a claim about quantum advantage, a hardware roadmap, or an algorithmic
speedup, and tell what it actually says.

By the end of the course you should understand core quantum mechanics principles, have Qiskit
installed and working, and be able to program a real quantum computer.

**Expected effort: 3-4 hours per week.**

### The Ethic of Mutual Aid

In each module it is important to ask and answer questions. All students should ask. Everyone
should try to answer at least once. We learn by solving our own problems and helping others solve
theirs.

This matters more in quantum than in most subjects, because the field is genuinely
counterintuitive. Nearly everyone hits a wall somewhere -- often at superposition, or at the
tensor product, or at the moment measurement stops feeling like reading a value and starts feeling
like a physical event. Hitting that wall is not a sign that you are behind. Saying so out loud is
usually what gets you past it, and it helps the next person who hits the same wall.

Mutual aid is a place to discuss the topic and the issue you are having. It is not a place to post
code, working or not, for feedback or solutions. Posting code is counterproductive in this type of
mutual aid environment.
