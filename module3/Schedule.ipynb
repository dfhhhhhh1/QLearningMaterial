# Module 3: Multiple Systems, Entanglement, and Interference

## Description

One qubit is a curiosity; many qubits are the point. This module extends the notation to composite systems via the tensor product, which is where the exponential scaling of the state space appears. Entanglement and interference -- the two resources that make quantum algorithms work -- are introduced here, along with quantum teleportation as the canonical worked example.

----


## Guiding Questions

What is quantum entanglement, how do these systems scale exponentially, and what does quantum teleportation mean?


## Topics Covered

- Tensor products and multi-qubit state vectors
- Exponential scaling of the state space
- Entanglement and the Bell states
- Interference as a computational resource
- Quantum teleportation
- Transpilation, shot count, simulator vs. real hardware


## Objectives

- Students will construct multi-qubit states using the tensor product
- Students will distinguish product states from entangled states
- Students will prepare a Bell state and interpret its measurement statistics
- Students will explain the role of interference in a quantum algorithm
- Students will describe what transpilation does and why shot count matters
