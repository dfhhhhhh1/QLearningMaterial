# Fundamentals

For all parts, global phase does not matter. That is, if the correct state is $|\phi\rangle$, then $e^{i\theta}|\phi\rangle$ for any $\theta$ is acceptable. In simpler terms, the $a$ in full state $a|\phi\rangle$ can be ignored if $|a| = 1$.

## Part 4: Measuring

You can add measurement gates ![](M3_E1_Quantum_Composer_Lab_media/media/image4.png) to your circuit to get outputs in the computational basis (binary). This will automatically cause the composer to run a simulation of your circuit. By default, it will run your circuit 1024 times and display the outputs of the measurement gates in the "Probabilities" graph at the bottom (or a dropdown to select a probabilities chart). You may hover over the bars in the graph to see how many times an output was observed.
