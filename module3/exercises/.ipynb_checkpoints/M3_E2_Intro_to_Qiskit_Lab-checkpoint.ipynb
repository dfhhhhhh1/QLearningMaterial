## Part 5: Simulation

We may now prepare our circuit to be ran on a backend. For now, we will just use a simulated backend. To do so we must first import a few more things.

You will want to install and import Qiskit Aer Simulator.


### INSTALL

```python
!pip install qiskit_aer
```


### IMPORT

```python
from qiskit_aer import AerSimulator
```

We first want to initialize AerSimulator.

```python
backend = AerSimulator()
```

The next step is to convert (transpile) your circuit into a circuit your backend can understand. Simulators are quite flexible and can take many gates as long as the qubit count is not too big. Real quantum computer backends usually only understand a small handful of native gates and circuit will need to be converted to those native gates. Luckily the transpiler does that for you. Sometimes your original circuit will already be compatible with the backend, in which case the transpiler is simply checking for compatibility.

We will also want to import the transpiler.

```python
from qiskit import transpile
```

Then we will want to transpile our circuit.

```python
transpiled_circuit = transpile(circuit, backend = backend)
```

The "TRANSPILED_CIRCUIT" is the new transpiled circuit that you may now run on your backend.

You may also display this circuit with the draw function to view how it has changed. From here we can send our circuit to our backend. We will need to first import the sampler. The sampler is one of the modes you can use to run circuits on a quantum computer. The sampler is the mode where you run a circuit then run measurement gates at the end and get binary results.

```python
from qiskit_ibm_runtime import SamplerV2 as Sampler
```

We can then run our circuit on our backend with the sampler. Use the Sampler constructor to initialize your sampler. Then use the run function to run your transpiled circuit on your backend. Be sure to specify your shot count.

```python
sampler = Sampler(mode = backend)
job = sampler.run([transpiled_circuit], shots = SHOT_COUNT)
```

You may then use the status function to check if your job has finished running.

```python
job.status()
```

Once it has finished running a simple way to obtain the results is to convert the job results into a dictionary.

```python
results = job.result()[0].data
outputs = {}
for key in results.__dict__:
    outputs[key] = results.__dict__[key].get_counts()
```

The above code will convert the results into a dictionary that can be printed out like below. The OUTPUTS dictionary will print out a list of possible outputs and how many times each output was observed.

```python
print(outputs)
```
