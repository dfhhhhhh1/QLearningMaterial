# Section 3: Encoding your data

This is one of the most important parts of the planning, how your data translates to a quantum state that can be evaluated by the QPU.

Before a quantum computer can do anything with your problem, the problem has to become qubits and
gates. How you do that determines how many qubits you need, how deep the circuit is, and therefore
whether it can run at all on present hardware.

## 3a: Compare the options

| Encoding | Idea | Qubits for $N$ values | Watch out for |
|---|---|---|---|
| **Basis** | each bitstring is one value | $N$ bits to a register | no compression; wide registers |
| **Amplitude** | data goes into the amplitudes of a state | $\log_2 N$ | exponentially compressed, but *preparing* the state can cost exponential depth |
| **Angle / rotation** | one value per rotation gate | 1 qubit per feature (or reuse) | shallow and hardware-friendly; limited capacity |
| **QUBO / Ising** | recast as minimizing a spin Hamiltonian | 1 qubit per binary variable | the standard route for optimization; constraints become penalty terms |
| **Hamiltonian mapping** | fermions or spins mapped to Paulis (Jordan-Wigner, parity) | problem dependent | the chemistry route; watch the Pauli term count |

***Pick at least two of these that could plausibly apply to your problem and compare them.*** *For
each: how many qubits would your real instance need, roughly how deep would the circuit be, and
what breaks first as you scale up?*

## 3b: The loading bottleneck

Classical data processing has the chance to reach scaling limits too.

If your problem involves a **large classical dataset**, ask how that data gets into the machine. An
encoding that needs $O(N)$ gates to load $N$ numbers has already spent the speedup before the
algorithm starts, you cannot get an exponential advantage if merely reading the input takes
exponential time. Amplitude encoding is the usual place this bites: the qubit count looks
wonderful, and the state-preparation circuit is the thing that kills you.

## 3d: Proof-of-concept encoding

Now make it concrete. Below is a runnable example of **angle encoding**: three data points turned into a rotation on its own qubit. If you find code/write your own and want to try another encoding method feel free.

***Adapt this cell to your own problem.*** *Replace the sample values with 2-3 real data points from
your project, adjust the encoding to match what you chose in 3c, and run it. If your chosen
encoding is QUBO/Ising, build a small* `SparsePauliOp` *for a 2-3 variable instance instead. If it
is basis encoding, prepare the bitstring.*

***Then state, in the markdown cell that follows, what the circuit represents and what would change
if you scaled it to your full problem size.***


```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# --- EXAMPLE: angle encoding of three feature values -------------------------
# Replace these with real values from YOUR problem.
raw_values = [0.20, 0.85, 0.55]          # e.g. three normalized sensor readings

# Angle encoding maps each value v in [0, 1] to a rotation angle in [0, pi].
angles = [np.pi * v for v in raw_values]

qc = QuantumCircuit(len(angles))
for qubit, angle in enumerate(angles):
    qc.ry(angle, qubit)

print(qc.draw())
print(f"\nqubits used: {qc.num_qubits}   circuit depth: {qc.depth()}")

# What state did we actually build? (Only feasible to print for tiny examples.)
state = Statevector(qc)
probs = state.probabilities_dict()
print("\nlargest basis-state probabilities:")
for bits, p in sorted(probs.items(), key=lambda kv: -kv[1])[:4]:
    print(f"  |{bits}>  {p:.4f}")
```
