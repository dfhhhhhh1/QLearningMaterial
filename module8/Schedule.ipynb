# Module 8: Current Research and Future Planning

## Description

The final module steps back from the tooling and looks at where the field is going and where you fit in it. Different hardware modalities make different bets; the logical-vs-physical qubit gap sets the timeline; and the question of which fields see value first is still genuinely open. The capstone is a quantum project plan for a problem of your choosing.

----


## Guiding Questions

Exploring different hardware, logical vs. physical qubits, and where and when people believe quantum will solve problems in different fields.


## Topics Covered

- Hardware modalities and their tradeoffs
- Logical vs. physical qubits and roadmap timelines
- Where quantum is expected to deliver value first, and why
- Becoming quantum ready
- Building a quantum project plan


## Objectives

- Students will compare quantum hardware modalities and articulate their tradeoffs
- Students will assess published roadmaps in terms of logical vs. physical qubit counts
- Students will evaluate a problem in their own domain for quantum suitability
- Students will produce a quantum project plan with a defensible scope and timeline
