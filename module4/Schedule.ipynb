# Module 4: Qiskit Programming and Circuits

## Description

This module turns the math into programs. You will map your existing intuition for classical logic gates onto quantum gates, similar in spirit, different in the details, because quantum gates are reversible and act on amplitudes. Using Qiskit you will build circuits, run them on simulators, and submit them to real quantum hardware. Deutsch's problem is the first algorithm where a quantum circuit provably beats the classical approach.

----


## Guiding Questions

Understand how to use your understanding of classical logic gates to program a quantum computer through similar but different "quantum gates." Also understand how to use Qiskit to test and run circuits on simulators and real quantum computers.


## Topics Covered

- Classical logic gates vs. quantum gates
- The standard gate set: X, Y, Z, H, S, T, CNOT
- Building circuits with Qiskit
- Simulators vs. real quantum computers
- Oracles and query models
- Deutsch's problem


## Objectives

- Students will construct quantum circuits in Qiskit from a gate-level description
- Students will run a circuit on both a simulator and real quantum hardware
- Students will explain what an oracle is and why the query model is used
- Students will implement and explain the circuit that solves Deutsch's problem
