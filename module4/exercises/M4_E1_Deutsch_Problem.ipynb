We know how X gates, Z gates, and CNOT gates behave in the computational basis $\left\{ \left| 0 \right\rangle,\left| 1 \right\rangle \right\}$. We will explore their behavior in the Hadamard basis $\left\{ \left| + \right\rangle,\left| - \right\rangle \right\}$. We can think of $\left| 0 \right\rangle$ mapping to $\left| + \right\rangle$ and $\left| 1 \right\rangle$ mapping to $\left| – \right\rangle$ through the H gate. We can encode a bit string such as 011010 into the Hadamard basis as $\left| + - - + - + \right\rangle$. You can also measure in the Hadamard basis by mapping $\left| + \right\rangle$ back to $\left| 0 \right\rangle$ and $\left| - \right\rangle$ back to $\left| 1 \right\rangle$ by using an H gate before a measurement gate. Where a 1 output can be interpreted as measuring $\left| - \right\rangle$, and a 0 output can be interpreted as measuring $\left| + \right\rangle$. Like so:

# Bonus: Deutsch's Algorithm

Knowing what we found out about CNOT gates in ***Section 1 Part 2: CNOT***, we will now modify our circuit to solve Deutsch's problem.

Deutsch's problem is as follows.

If we have some function $f$ that takes a 1-bit input and a 1-bit output, how can we efficiently tell if the function is balanced or constant?

That is to say, how can we know if the function outputs the same thing no matter what the input ($f_{1}(x)$ and $f_{4}(x)$) is (constant). Or if it outputs a 1 for 50% of the inputs and a 0 for the other 50% of the inputs ($f_{2}(x)$ and $f_{3}(x)$) (balanced).

Reminder of our 4 Deutsch oracles.

The four possible one-bit oracles ($f_1$ and $f_4$ are **constant**; $f_2$ and $f_3$ are **balanced**):

| $x$ | $f_1(x)$ | $f_2(x)$ | $f_3(x)$ | $f_4(x)$ |
|:---:|:---:|:---:|:---:|:---:|
| $0$ | $0$ | $0$ | $1$ | $1$ |
| $1$ | $0$ | $1$ | $0$ | $1$ |

In this problem, we assume that we do not know which oracle is being used, we can only give inputs and measure outputs. On a classical computer the most efficient way to tell if an oracle is balanced or constant is to try both inputs 1 and 0. If they give the same output then the oracle is constant, if they give different outputs then it is balanced. This means we have to run the oracle 2 times. On a quantum computer, we can solve this problem by only running the oracle 1 time.

Since the oracle is now off limits, we must solve this problem by only modifying the things outside of the oracle. In the circuit below, the only gates we may modify are the "Set inputs" and" "Process Outputs" gates. in other words, choose an input, and interpret the output as we may please.
